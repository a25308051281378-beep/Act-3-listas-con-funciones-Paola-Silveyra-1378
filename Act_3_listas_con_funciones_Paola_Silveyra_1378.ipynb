#### Listas con funciones while true

```python
# Ejemplo de condicional con "bailarín"

profesion = "bailarín"

if profesion == "bailarín":
    print("Eres bailarín.")
else:
    print("Profesión no encontrada.")
```

```python
# Ejemplo con entrada de usuario

profesion = input("Escribe una profesión: ")

if profesion == "bailarín":
    print("Eres bailarín.")
else:
    print("Profesión no encontrada.")
```

```python
# Ejemplo con varias opciones

profesion = input("Escribe una profesión: ")

if profesion == "bailarín":
    print("Te gusta bailar.")

elif profesion == "cantante":
    print("Te gusta cantar.")

else:
    print("Profesión no registrada.")
```

```python
# Menú simple con while True

while True:

    print("===== MENÚ =====")
    print("1. Ver profesión")
    print("2. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":

        profesion = input("Escribe la profesión: ")

        if profesion == "bailarín":
            print("Eres bailarín.\n")

        else:
            print("Profesión no encontrada.\n")

    elif opcion == "2":

        print("Programa finalizado.")
        break

    else:
        print("Opción inválida.\n")


        print("Programa desrrollado por Paola Silveyra nl=50")
```
