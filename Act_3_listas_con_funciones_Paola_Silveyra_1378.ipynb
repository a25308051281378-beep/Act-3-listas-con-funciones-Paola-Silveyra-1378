#### Listas con funciones while true

```python
# ==========================================
# FUNCIÓN agregarProfesion()
# ==========================================

def agregarProfesion():

    profesion = "bailarín"

    print("Profesión agregada:", profesion)


# Ejecutar función
agregarProfesion()
```

```python
# ==========================================
# FUNCIÓN leerProfesion()
# ==========================================

def leerProfesion():

    profesion = "bailarín"

    print("La profesión es:", profesion)


# Ejecutar función
leerProfesion()
```

```python
# ==========================================
# FUNCIÓN listarProfesiones()
# ==========================================

def listarProfesiones():

    profesiones = ["bailarín", "cantante", "actor"]

    print("Lista de profesiones:")

    for profesion in profesiones:
        print(profesion)


# Ejecutar función
listarProfesiones()
```

```python
# ==========================================
# FUNCIÓN buscarProfesion()
# ==========================================

def buscarProfesion():

    profesion = "bailarín"

    buscar = "bailarín"

    if profesion == buscar:
        print("Profesión encontrada.")

    else:
        print("Profesión no encontrada.")


# Ejecutar función
buscarProfesion()
```

```python
# ==========================================
# FUNCIÓN borrarProfesion()
# ==========================================

def borrarProfesion():

    profesion = "bailarín"

    print("Profesión eliminada:", profesion)

    profesion = ""

    print("Registro vacío.")


# Ejecutar función
borrarProfesion()
```

```python
# ==========================================
# MENÚ CON while True
# ==========================================

def agregar():
    print("Profesión agregada: bailarín\n")


def leer():
    print("Profesión registrada: bailarín\n")


def listar():
    print("bailarín")
    print("cantante")
    print("actor\n")


def buscar():
    print("Profesión encontrada: bailarín\n")


def modificar():
    print("Profesión modificada correctamente\n")


def borrar():
    print("Profesión eliminada correctamente\n")


while True:

    print("===== MENÚ =====")
    print("1. Agregar")
    print("2. Leer")
    print("3. Listar")
    print("4. Buscar")
    print("5. Modificar")
    print("6. Borrar")
    print("7. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        agregar()

    elif opcion == "2":
        leer()

    elif opcion == "3":
        listar()

    elif opcion == "4":
        buscar()

    elif opcion == "5":
        modificar()

    elif opcion == "6":
        borrar()

    elif opcion == "7":
        print("Programa finalizado.")
        break

    else:
        print("Opción inválida.\n")

        print("Programa desarrollado por Paola Silveyra NL = 50bailarín")
```
